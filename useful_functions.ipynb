## Generate one big csv file with all the low level features

```python
import pandas as pd
import glob
import os
import re
```

```python
folder_path = "C:/Users/vassa/Desktop/UZH/Masters Project/synthetic_network_data_gen/test_gan_data/quiche"

csv_files = glob.glob(os.path.join(folder_path, "*.csv"))

dfs = []

for file in csv_files:
    match = re.search(r'quiche_capture_(\d+)', os.path.basename(file))
    capture_id = int(match.group(1)) if match else None

    df = pd.read_csv(file)

    df["capture_id"] = capture_id

    dfs.append(df)
merged_df = pd.concat(dfs, ignore_index=True)

merged_df.to_csv(os.path.join(folder_path, "merged_captures_1_30.csv"), index=False)

print("✅ Merged CSV created successfully!")
```

## Generate one big csv files using the csv files from a zipped folder

```python
import os
import re
import pandas as pd
import zipfile


zip_file_path = "C:/Users/vassa/Desktop/UZH/Masters Project/synthetic_network_data_gen/test_gan_data/low_level_features/quiche.zip"

output_folder_path = "C:/Users/vassa/Desktop/UZH/Masters Project/synthetic_network_data_gen/test_gan_data/low_level_features"


dfs = []

with zipfile.ZipFile(zip_file_path, 'r') as zf:
    csv_files = [f for f in zf.namelist() if f.endswith('.csv')]

    for file in csv_files:
        match = re.search(r'quiche_capture_(\d+)', os.path.basename(file))
        capture_id = int(match.group(1)) if match else None
        
        with zf.open(file) as f:
            df = pd.read_csv(f)

        df["capture_id"] = capture_id
        dfs.append(df)

# Concatenate all DataFrames into one
merged_df = pd.concat(dfs, ignore_index=True)

# Save the merged DataFrame to a new CSV file
merged_df.to_csv(os.path.join(output_folder_path, "merged_captures_1_30.csv"), index=False)

print("✅ Merged CSV created successfully!")
```

## Rename capture files by ofsetting the number

```python
import os
import re


# Folders
source_folder = r"C:\Users\vassa\Desktop\UZH\Masters Project\synthetic_network_data_gen\captures\pcap_files\quicgo\quicgo"

# Start numbering from this value
offset = 0   # if your other folder already goes up to 3000

pattern = re.compile(r"^(\d+)_(.+)$")

for filename in os.listdir(source_folder):
    match = pattern.match(filename)
    if match:
        old_number = int(match.group(1))
        new_number = old_number + offset
        rest = match.group(2)   # everything after the first number_

        old_path = os.path.join(source_folder, filename)
        new_name = f"{new_number}_{rest}"
        new_path = os.path.join(source_folder, new_name)

        os.rename(old_path, new_path)
        print(f"Renamed {filename} → {new_name}")
```
